1. Create a 3x3 array with random integers between 1 and 100. Write a program to
"normalize" the array so that the values are scaled between 0 and 1.

    • Formula: (x - min) / (max - min)
    
    • Example Output: An array where the smallest value becomes 0.0 and the
largest becomes 1.0.

````python
import numpy as np

arr = np.random.randint(1, 101, size=(3, 3))

arr_min = arr.min()
arr_max = arr.max()

arr_normalized = (arr - arr_min) / (arr_max - arr_min)

print("Normalized array:")
print(arr_normalized)
````

2. Create a 1D array of 10 random numbers. Ask the user for a value X Find the value
in the array that is closest to X

    • Example Input: Array: [1, 5, 8, 12], X = 10

    • Example Output: 8 or 12 (whichever is mathematically closer)

````python
import numpy as np

arr = np.random.randint(1, 21, size=10)
print("Array:", arr)

X = int(input("Enter a value X: "))

below = arr[arr <= X]  
above = arr[arr >= X]   

if below.size == 0:
    print(f"The closest value to {X} is: {above.min()}")
elif above.size == 0:
    print(f"The closest value to {X} is: {below.max()}")
else:
    lower = below.max()  
    upper = above.min()  

    dist_lower = X - lower   
    dist_upper = upper - X   

    if dist_lower < dist_upper:
        print(f"The closest value to {X} is: {lower}")
    elif dist_upper < dist_lower:
        print(f"The closest value to {X} is: {upper}")
    else:
        if lower == upper:
            print(f"The closest value to {X} is: {lower}")
        else:
            print(f"The closest values to {X} are: {lower} and {upper}")
````

3. Write a NumPy program to create a 5x5 array with random values. For each column
in the array, find the second-largest value.

    • Example Input: [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

    • Example Output: [4, 5, 6] (These are the second-highest values for each
respective column).

````python
import numpy as np

arr = np.random.randint(1, 21, size=(5, 5))
print("Original array:")
print(arr)

sorted_arr = np.sort(arr, axis=0)

second_largest = sorted_arr[-2]

print("\nSecond-largest value in each column:")
print(second_largest)
````

4. Create a 6x4 matrix of random integers. Identify which rows have a sum that is an
even number. Print the indices of these rows.


````python
import numpy as np

arr = np.random.randint(1, 21, size=(6, 4))
print(arr)

row_sums = arr.sum(axis=1)
print("Row sums:", row_sums)

even_rows = []
for i in range(len(row_sums)):
    if row_sums[i] % 2 == 0:
        even_rows.append(i)

print("Indices of rows with even sum:", even_rows)
````

5. Create a 1D array of 30 random integers between 1 and 50. Replace all numbers
that are multiples of 3 or 5 with the value -1.

    • Example Input: [3, 7, 10, 15]

    • Example Output: [-1, 7, -1, -1]

````python
import numpy as np

arr = np.random.randint(1, 51, size=30)
print(arr)

arr[(arr % 3 == 0) | (arr % 5 == 0)] = -1

print("Array after replace:")
print(arr)
````

6. Write a NumPy program to create a 3x3 array with random values. For each row,
calculate its mean (average) and subtract that mean from every element in that
specific row.

    • Example Input: ``` [[10, 20, 30], [5, 5, 5], [0, 10, 20]]

    • Example Output: [[-10. 0. 10.], [ 0. 0. 0.], [-10. 0. 10.]]


````python
import numpy as np

arr = np.random.randint(1, 31, size=(3, 3))
print(arr)

result = np.zeros((3, 3))

result = arr - arr.mean(axis=1, keepdims=True)

print("After subtracting row means:")
print(result)
````
